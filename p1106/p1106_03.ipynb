```python
from sklearn import svm,metrics
import pandas as pd
```

```python
df = pd.read_csv('data/iris.csv',index_col='caseno')
# df.shape
# df.columns
# df.info() 
df['Species'].unique()
```

```python
df.columns
```

```python
# 1. 데이터준비
# - 전체데이터를 가져옴 
# - 정답컬럼 문자를 숫자로 변경
# - 데이터 분리 - data:학습,label:정답 분리
# - 훈련데이터,테스트데이터 분리 (8:2)
# 2. 데이터학습 fit
# 3. 데이터예측 predict
# 4. 데이터검증 score
```

```python
# 데이터 준비
# 150개 - str컬럼을 변경해야 함.
#df['Species'].unique()
def nameChange(s):
    if s == 'Iris-setosa':
        return 0
    elif s == 'Iris-versicolor':
        return 1
    else:          # Iris-virginica
        return 2
df['Species'] = df['Species'].apply(nameChange)
df.head()
```

```python
# 데이터 분리 - data,label
# x_data,x_test,y_data,y_test
# x_train,x_test,y_train,y_test
# train_data,test_data,train_label,test_label
# 데이터의 학습 - 150
data = df[['SepalLength','SepalWidth','PetalLength','PetalWidth']]
# 데이터의 정답 - 150
label = df['Species']
data.head(3)
```

```python
# 150개 데이터에서 학습데이터, 테스트데이터를 분리
# 8:2 비율로 학습데이터와 테스트데이터를 분리

# 학습데이터, 테스트데이터 분리 함수
# train_test_split() 함수 사용
from sklearn.model_selection import train_test_split
# test_size=0.2 : 20%를 테스트데이터로 분리
# stratify=label : label의 비율을 유지하면서 분리
train_data,test_data,train_label,test_label = \
    train_test_split(data,label, test_size=0.2, stratify=label,\
                     random_state=0)
```

```python
# 2. 데이터학습 fit
clf = svm.SVC()
clf.fit(train_data,train_label) # 120개 학습데이터로 학습
```

```python
# 3. 데이터예측 predict
result = clf.predict(test_data) # 30개 테스트데이터로 예측
print("예측 결과 : ", result)
```

```python
result2 = result.tolist() # 리스트타입으로 변경
```

```python
def nameChange2(s):
    if s == 0:
        return 'Iris-setosa'
    elif s == 1:
        return 'Iris-versicolor'
    else:          # Iris-virginica
        return 'Iris-virginica'

result_names = [nameChange2(s) for s in result2]  # 리스트 for문으로 변경
result_names
```

```python
# 4. 데이터검증 score
# score = metrics.accuracy_score(test_label,result)
score = clf.score(test_data,test_label) # 30개 테스트데이터로 검증
print("정답률 : ", score)
```
