```python
df = pd.read_csv('data/iris.csv')
df.head(2)
```

```python
df['Species'].unique()
```

```python
label = df['Species'].values
data = df.iloc[:,1:-1].values
data[:2]
```

```python
train_data,test_data,train_label,test_label = \
train_test_split(data,label,random_state=42)
```

```python
ss = StandardScaler()
ss.fit(train_data)
train_scaled = ss.transform(train_data)
test_scaled = ss.transform(test_data)
```

```python
train_score = []
test_score = []
cc = [0.001,0.01,0.1,1,10,100]
for c in cc:
    lr = LogisticRegression(C=c, max_iter=1000)
    lr.fit(train_scaled,train_label)
    train_score.append(lr.score(train_scaled,train_label))
    test_score.append(lr.score(test_scaled,test_label))
    print("c값",c)
    print(f"train정확도: {(train_score[-1]):.3f}")
    print(f"test정확도: {(test_score[-1]):.3f}")
```

```python
indata = input('꽃잎 길이, 너비, 꽃받침 길이, 너비를 입력하세요.(예:5.1,3.5,1.4,0.2)')
indata = indata.split(',')
indata = np.array(indata).astype(float).reshape(1,-1)
```

```python
indata_scaled = ss.transform(indata)
```

```python
predict = lr.predict(indata_scaled)
predict
```

```python
proba = lr.predict_proba(indata_scaled)
proba[0][0]
```
