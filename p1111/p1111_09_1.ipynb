```python
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams['font.family'] = 'Malgun Gothic'
matplotlib.rcParams['axes.unicode_minus']=False

import pandas as pd
import numpy as np
from sklearn import svm,metrics
from sklearn.neighbors import KNeighborsClassifier
from sklearn.neighbors import KNeighborsRegressor
from sklearn.linear_model import LinearRegression 
from sklearn.preprocessing import PolynomialFeatures  # 다항특성 자동생성
from sklearn.linear_model import Ridge
from sklearn.linear_model import Lasso
from sklearn.metrics import mean_absolute_error   # 평균절대값오차 오차범위
from sklearn.preprocessing import StandardScaler  # 정규화클래스
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv('data/manhattan.csv')
```

```python
label = df['rent'].values
data = df.drop(columns=['rental_id','rent','neighborhood','borough']).values
```

```python
train_data, test_data, train_label, test_label = \
train_test_split(data,label)
```

```python
lr = LinearRegression()
lr.fit(train_data,train_label)
```

```python
predict = lr.predict(test_data)
print("예측치:",predict)
score = lr.score(train_data,train_label)
score2 = lr.score(test_data,test_label)
print("train정확도", score)
print("test정확도", score2)
```

```python
# 자료분리는 이미 완료
# 다항식으로 변환
poly = PolynomialFeatures(degree=2, include_bias=False)
poly.fit(train_data)
train_poly = poly.transform(train_data)
test_poly = poly.transform(test_data)
```

```python
# 정규화
ss = StandardScaler()
ss.fit(train_poly)
train_scaled = ss.transform(train_poly)
test_scaled = ss.transform(test_poly)
```

```python
alpha_list = [0.001,0.01,0.1,1,10,100,1000]
for alpha in alpha_list:
    ridge = Ridge(alpha=alpha)
    ridge.fit(train_scaled,train_label)
    score = ridge.score(train_scaled,train_label)
    score2 = ridge.score(test_scaled,test_label)
    print("alpha값:",alpha)
    print("train정확도",score)
    print("test정확도",score2)
# alpha판정 : train값이 높고, test값은 train값보다는 낮지만, 다른 alpha값에 따른 가장 높은 test값
```
